Internship project for Sanket Shantaram Tarle. The notebook demonstrates data loading, EDA, preprocessing, model training, evaluation and prediction. The bundled CSV is a synthetic offline demonstration dataset following the variable style of the UCI Student Performance dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import joblib
```

```python
df = pd.read_csv('../data/student_performance.csv')
df.head()
```

```python
print('Shape:', df.shape)
display(df.describe(include='all').T)
print('Missing values:\n', df.isnull().sum())
```

```python
plt.figure(figsize=(7,4))
plt.hist(df['G3'], bins=range(0,22), edgecolor='black')
plt.title('Final Grade Distribution')
plt.xlabel('G3'); plt.ylabel('Students'); plt.show()
```

```python
plt.figure(figsize=(7,4))
plt.scatter(df['G2'], df['G3'], alpha=0.55)
plt.title('G2 vs G3')
plt.xlabel('G2'); plt.ylabel('G3'); plt.show()
```

```python
X = df.drop(columns=['G3'])
y = df['G3']
cat = [c for c in X.columns if X[c].dtype == 'object']
num = [c for c in X.columns if c not in cat]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
pre = ColumnTransformer([('num', StandardScaler(), num), ('cat', OneHotEncoder(handle_unknown='ignore'), cat)])
model = Pipeline([('preprocessor', pre), ('regressor', RandomForestRegressor(n_estimators=300, max_depth=10, min_samples_leaf=2, random_state=42))])
model.fit(X_train, y_train)
pred = model.predict(X_test)
print('MAE:', mean_absolute_error(y_test, pred))
print('RMSE:', mean_squared_error(y_test, pred)**0.5)
print('R2:', r2_score(y_test, pred))
```

```python
joblib.dump(model, '../models/student_performance_model.joblib')
print('Model saved.')
```
